```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

from scipy.stats import ks_1samp,norm,uniform
from scipy import stats
import pandas as pd
```

```python
import os
current_dir = os.getcwd()
current_dir
```

```python
file_path = "/path/to/replication_folder/1_data/5_meta/4_pval_other"
```

```python
# TMF
df = pd.read_stata(f"{file_path}/tmf.dta")

# List comprehension to perform KS tests for each measure
KS_pval = pd.DataFrame([
    {
        'measure': measure,
        'p_value': round(ks_1samp(df[df['measure'] == measure].p_val, uniform.cdf, alternative='greater')[1], 3)
    }
    for measure in ["tmf", "CGI"]
])

KS_pval.to_stata(f"{file_path}/tmf_KS.dta", version=117)
```
