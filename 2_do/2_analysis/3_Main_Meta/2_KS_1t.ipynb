```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

import scipy
from scipy.stats import ks_1samp,norm,uniform
from scipy import stats
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
file_path = "/path/to/replication_folder/1_data/5_meta/1_pval"
```

```python
# Reading the Stata file into a DataFrame
dfs = []  # List to store DataFrames

for i in range(3, 8):
    df = pd.read_stata(f"{file_path}/table_{i}_panelC.dta")
    dfs.append(df)

result_df = pd.concat(dfs, ignore_index=True)
result_df = result_df[result_df['dropped_panelC'] != 1]

ks_p_value=round(ks_1samp(result_df.p_c_cgi_1t,uniform.cdf, alternative='greater')[1], 3)
ks_df = pd.DataFrame({'ks_p_value': [ks_p_value]})
ks_df.to_stata(f"{file_path}/CGI_pooled_KS.dta")
```
